```python
import gensim
import os
```

```python
from nltk import sent_tokenize
from nltk.corpus import stopwords
from gensim.utils import simple_preprocess

stop_words = set(stopwords.words('english'))

story = []
for filename in os.listdir('data'):
    
    f = open(os.path.join('data',filename),  encoding='cp1252') # this encoding needed due to not using of the utf-8 encoding 
    corpus = f.read()
    raw_sent = sent_tokenize(corpus)
    for sent in raw_sent:
        tokens = simple_preprocess(sent)

        tokens = [
            word for word in tokens
            if word not in stop_words # avoid ing the stop words
        ]

        story.append(tokens)
    
```

```python
story
```

```python
model = gensim.models.Word2Vec(
    window=10, # size of the sliding windows
    min_count=2,  # sentences with at least two words is used
    # workers=4 no of cores to use
)
```

```python
model.build_vocab(story) # building the vocab 
```

```python
# Training 
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
X = pca.fit_transform(model.wv.get_normed_vectors())
```
